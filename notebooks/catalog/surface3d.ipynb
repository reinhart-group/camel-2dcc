# Real microscope scans you can spin

Three scans, each a still picture until you tap **spin this one**. Tapping one puts the previous one back to a picture.

That is not a stylistic choice. Safari cannot keep more than one live 3D view on a page here: with three at once, every one of them goes blank. One at a time works perfectly, so the page only ever holds one.

The still pictures are real renders of the same data, so nothing is lost if you never tap, if the 3D library cannot be reached, or if the browser reclaims the view.
