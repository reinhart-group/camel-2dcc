# What can this device actually draw?

You saw colour bars but no surface. Colour bars are ordinary SVG; the surface itself needs **WebGL**. So Plotly loaded and ran, and only the 3D part failed — which means my last two theories were both wrong.

You also said the second plot drew once and the first did not. That is the signature of a **WebGL context limit**: a browser holds only a few live 3D canvases at a time and throws away the oldest, so the newest plot survives and earlier ones go blank.
